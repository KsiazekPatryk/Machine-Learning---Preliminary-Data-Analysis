```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error

import numpy as np
from pathlib import Path
```

```python
data_path = Path("/Users/patrykksiazek/Desktop/Machine-Learning---Preliminary-Data-Analysis/")
```

```python
train_data = pd.read_csv(data_path / "train_data.csv", index_col=0)
test_data = pd.read_csv(data_path / "test_data.csv", index_col=0)
```

```python
#Separating features and target variable
X_train = train_data.drop("score", axis=1)
y_train = train_data["score"]
X_test = test_data.drop("score", axis=1)
y_test = test_data["score"]
```

```python
def train_and_evaluate(model, X_train=X_train, X_test=X_test):
    """ Trenuje model regresji liniowej i oblicza RMSE na zbiorach treningowych i testowym"""
    model.fit(X_train,y_train)

    #Przewidywanie i obliczanie RMSE
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
    rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))

    return rmse_train, rmse_test,y_test_pred
```

```python
scaler = StandardScaler()
```

```python
X_train_normalized = scaler.fit_transform(X_train)
X_test_normalized = scaler.transform(X_test)
```

```python
nn_regressor = MLPRegressor(max_iter=1000, early_stopping=True)
```

```python
train_and_evaluate(nn_regressor, X_train_normalized, X_test_normalized)[:2]
```
